```python
import pandas as pd
df = pd.read_csv('../data/bobo_benz_workshop_data_1000.csv')
```

```python
print(list(df.columns))
```

```python
mapeamento_binario = {
    'Yes': True,
    'YES': True,
    'yes': True,
    'Y': True,
    'y': True,
    'No': False,
    'NO': False,
    'no': False,
    'N': False,
    'n': False,
}

colunas_binarias = [
    'Has_Insurance_Cover',
    'Request_For_Pickup',
    'Home_Service',
    'Towing_Required',
    'Whatsapp_Followup',
]

for coluna in colunas_binarias:
  if coluna in df.columns:
    df[coluna] = df[coluna].map(mapeamento_binario).fillna(False)


df = df.where(df.notna(), None)
```

```python
df['Date'] = pd.to_datetime(df['Date'], format='mixed', dayfirst=True, errors='coerce')
```

```python
colunas_texto = df.select_dtypes(include=['string']).columns
for col in colunas_texto:
    df[col] = df[col].str.strip().str.capitalize()
```

```python
df = df.replace('None', '')
```
