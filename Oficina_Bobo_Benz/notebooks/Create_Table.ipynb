```python
import mysql.connector
import pandas as pd

conexao = mysql.connector.connect(
    host="localhost",
    user="root",
    password="*********",
    database="bobo_benz"
)

cursor = conexao.cursor()
```

```python

cursor.execute("""
CREATE TABLE IF NOT EXISTS Clients (
    Customer_ID VARCHAR(20) PRIMARY KEY,
    Customer_Name VARCHAR(100),
    Phone VARCHAR(20),
    Customer_Type VARCHAR(50),
    Whatsapp_Followup BOOLEAN,
    Fleet_Name VARCHAR(50)
)
""")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS Vehicle (
    Vehicle_ID INT PRIMARY KEY,
    Customer_ID VARCHAR(20),
    Vehicle_Type VARCHAR(50),
    Plate_Number VARCHAR(20) UNIQUE,
    Has_Insurance_Cover BOOLEAN,
    FOREIGN KEY (Customer_ID) REFERENCES Clients(Customer_ID)
)
""")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS Mechanic (
    Mechanic_ID INT PRIMARY KEY,
    Mechanic_Name VARCHAR(100),
    Skill_Level VARCHAR(50)
)
""")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS Material (
    Material_ID INT PRIMARY KEY,
    Material_Name VARCHAR(50)
)
""")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS Services (
    Service_ID INT PRIMARY KEY,
    Customer_ID VARCHAR(20),
    Vehicle_ID INT,
    Mechanic_ID INT,
    Material_ID INT,
    Service_Date DATE, 
    Service_Type VARCHAR(50),
    Complaint TEXT,
    Status VARCHAR(50),
    Service_Time_Hours DECIMAL(5,2),
    Wait_Time_Mins INT,
    Service_Cost DECIMAL(10,2),
    Parts_Cost DECIMAL(10,2),
    Labour_Cost DECIMAL(10,2),
    Discount_Given DECIMAL(10,2),
    Rating DECIMAL(3,1),
    Payment_Mode VARCHAR(50),
    Referral_Source VARCHAR(50),
    Workshop_Location VARCHAR(50),
    Parts_Source VARCHAR(50),
    Spare_Parts_Availability VARCHAR(50),
    Service_Priority VARCHAR(50),
    Request_For_Pickup BOOLEAN,
    Home_Service BOOLEAN,
    Towing_Required BOOLEAN,
    Service_Package VARCHAR(50),
    Return_Visit BOOLEAN,
    Promo_Code_Used VARCHAR(50),
    Follow_Up_Date DATE,
    Mechanic_Notes TEXT,
    Customer_Notes TEXT,
    Remarks TEXT,
    FOREIGN KEY (Customer_ID) REFERENCES Clients(Customer_ID),
    FOREIGN KEY (Vehicle_ID) REFERENCES Vehicle(Vehicle_ID),
    FOREIGN KEY (Mechanic_ID) REFERENCES Mechanic(Mechanic_ID)
)
""")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS ServicesMaterial (
    Service_ID INT,
    Material_ID INT,
    Quantity_Used INT, -- Adicionado para saber quantas peças foram usadas
    PRIMARY KEY (Service_ID, Material_ID),
    FOREIGN KEY (Service_ID) REFERENCES Services(Service_ID),
    FOREIGN KEY (Material_ID) REFERENCES Material(Material_ID)
)
""")
conexao.commit()
print("Tabelas criadas com sucesso!")
cursor.close()
conexao.close()
```
